```python
# Task 1: Interactive introduction
name = input("Enter your name: ")
age = input("Enter your age: ")
favourite_colour = input("Enter your favourite colour: ")

print(f"Hello {name}! You are {age} years old, and your favourite colour is {favourite_colour}.")
```

```python
# Task 2: Rectangle calculator
length = float(input("Enter the length of the rectangle: "))
width = float(input("Enter the width of the rectangle: "))

area = length * width
perimeter = 2 * (length + width)

print(f"Area: {area}")
print(f"Perimeter: {perimeter}")
```

```python
# Task 3: Simple shopping calculator
product_name = input("Enter product name: ")
unit_price = float(input("Enter unit price: "))
quantity = int(input("Enter quantity: "))

total_cost = unit_price * quantity
print(f"Receipt -> Product: {product_name} | Unit Price: ₦{unit_price:.2f} | Qty: {quantity} | Total: ₦{total_cost:.2f}")
```

```python
# Task 4: String methods practice
sentence = input("Enter a sentence: ")

print("Uppercase:", sentence.upper())
print("Lowercase:", sentence.lower())
print("Title Case:", sentence.title())
print("Number of characters:", len(sentence))
```

```python
# Task 5: Personal profile
full_name = input("Enter your full name: ")
city = input("Enter your city: ")
hobby = input("Enter a hobby: ")

clean_name = full_name.strip()
clean_city = city.strip()

print(f"Profile Summary:\nName: {clean_name}\nCity: {clean_city}\nHobby: {hobby}")
```

```python
# Temperature converter
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9 / 5) + 32
print(f"{celsius}°C is equal to {fahrenheit}°F")
```

```python
# Extension challenge
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print(f"Sum: {num1 + num2}")
print(f"Difference: {num1 - num2}")
print(f"Product: {num1 * num2}")

if num2 == 0:
    print("Division and remainder cannot be calculated because the second number is zero.")
else:
    print(f"Quotient: {num1 / num2}")
    print(f"Remainder: {num1 % num2}")
```
