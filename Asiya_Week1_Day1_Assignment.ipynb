```python
# Task 1 About me
first_name = "Salma"
age = 25
country = "Nigeria"
is_student = True

print("First Name:", first_name)
print("Age:", age)
print("Country:", country)
print("Current Student:", is_student)
```

```python
# Task 2 Data type detective
whole_num = 42
decimal_num = 3.14
text_val = "Hello Asiya"
bool_val = True

print(type(whole_num))
print(type(decimal_num))
print(type(text_val))
print(type(bool_val))
```

```python
# Task 3 - Calculate your age in months
age_years = 25
age_months = age_years * 12
print(f"If your age in years is {age_years}, your age in months is approximately {age_months} months.")
```

```python
# Task 4 Fix the conversion
age_str = "22"
height_str = "1.65"

age = int(age_str)
height = float(height_str)

print("Converted Age:", age, "Type:", type(age))
print("Converted Height:", height, "Type:", type(height))
```

```python
# Task 5 Mini challenge: Shopping total
item_price = 1500.50
quantity = 3
total_cost = item_price * quantity

print(f"Item Price: ₦{item_price} | Quantity: {quantity}")
print(f"Total Cost: ₦{total_cost}")
```

```python
# Extension challenge
student_name = "Zara"
score1 = 85.5
score2 = 90.0
score3 = 78.5

average_score = (score1 + score2 + score3) / 3

print(f"Student Name: {student_name}")
print(f"Average Score: {average_score:.2f}")
```
